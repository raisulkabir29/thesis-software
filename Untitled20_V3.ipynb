# VERIFIED V3 — Three-Source Agricultural Production Model

This is the three-source version. It trains with `training_data.csv`, evaluates with `test_data.csv`, and retrieves district names from `full_agriculture_data.csv`.

```python
from thesis_model_v3 import (
    FEATURE_COLUMNS,
    FULL_AGRICULTURE_FILE,
    PROGRAM_VERSION,
    TEST_DATA_FILE,
    TRAINING_DATA_FILE,
    build_district_code_lookup,
    build_results_table,
    evaluate_model,
    load_district_names,
    load_model_data,
    predict_production,
    prediction_frame,
    train_model,
    validate_three_data_sources,
)

print(f"Loaded program version: {PROGRAM_VERSION}")
```

## 1. Load all three CSV files

```python
training_data = load_model_data(TRAINING_DATA_FILE)
test_data = load_model_data(TEST_DATA_FILE)
district_names = load_district_names(FULL_AGRICULTURE_FILE)

validate_three_data_sources(training_data, test_data, district_names)

print(f"Training source: {TRAINING_DATA_FILE} — {len(training_data)} rows")
print(f"Test source: {TEST_DATA_FILE} — {len(test_data)} rows")
print(f"Name source: {FULL_AGRICULTURE_FILE} — {len(district_names)} rows")
```

## 2. Train with training_data.csv and evaluate with test_data.csv

```python
model = train_model(training_data)
metrics = evaluate_model(model, test_data)

{
    "R-squared": metrics["r2"],
    "MAE": metrics["mae"],
    "RMSE": metrics["rmse"],
    "WMAPE": metrics["wmape"],
    "Median absolute percentage error": metrics["median_ape"],
}
```

## 3. Final test results with district names

The predictions below come from `test_data.csv`. `District_Name` is matched from `full_agriculture_data.csv` by record ID. Numeric IDs and district codes are not displayed.

```python
results = build_results_table(model, test_data, district_names)
results.head(20)
```

## 4. Predict a new record using a district name

```python
district_codes = build_district_code_lookup(
    training_data, test_data, district_names
)
new_record = prediction_frame(
    ["Bandarban", 900, 2488, 4, 22, 486, 47.093, 0.248, 1122.322, 2018],
    district_codes,
)
prediction = predict_production(model, new_record)[0]
print("District: Bandarban")
print(f"Predicted production: {prediction:.2f} M.Tons")
```
